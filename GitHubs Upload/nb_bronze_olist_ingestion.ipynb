```python
# Dictionary: left side = CSV file name, right side = Bronze table name we want to create
files = {
    "olist_orders_dataset.csv": "bronze_orders",                      # orders -> bronze_orders
    "olist_customers_dataset.csv": "bronze_customers",                # customers -> bronze_customers
    "olist_order_items_dataset.csv": "bronze_order_items",            # items in each order
    "olist_order_payments_dataset.csv": "bronze_order_payments",      # payment details
    "olist_order_reviews_dataset.csv": "bronze_order_reviews",        # customer reviews
    "olist_products_dataset.csv": "bronze_products",                  # product details
    "olist_sellers_dataset.csv": "bronze_sellers",                    # seller details
    "olist_geolocation_dataset.csv": "bronze_geolocation",            # zip code to lat/long
    "product_category_name_translation.csv": "bronze_category_translation",  # Portuguese -> English category names
}

# Loop: repeat the steps below once for every file in the dictionary above
for file_name, table_name in files.items():
    df = (                                   # df = DataFrame, a table held in memory by Spark
        spark.read                           # tell Spark we want to read data
        .option("header", True)              # first row of the CSV contains column names
        .option("inferSchema", True)         # let Spark guess data types (number, text, date)
        .option("multiLine", True)           # allow text fields that span multiple lines (needed for reviews)
        .option("quote", '"')                # text fields are wrapped in double quotes
        .option("escape", '"')               # a double quote inside text is written as two quotes
        .csv(f"Files/olist/{file_name}")     # read the CSV from the Files/olist folder of the lakehouse
    )
    df.write.mode("overwrite").format("delta").saveAsTable(table_name)  # save as a Delta table (overwrite if it already exists)
    print(f"{table_name}: {df.count()} rows")  # print the table name and how many rows it has
```
